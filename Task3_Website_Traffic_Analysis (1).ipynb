```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
traffic = pd.read_csv('traffic.csv')
traffic.head()
```

```python
print("Rows and Columns:", traffic.shape)
traffic.info()
```

```python
traffic.isnull().sum()
```

```python
traffic.fillna('Unknown', inplace=True)
traffic.isnull().sum()
```

```python
total_sessions = traffic['linkid'].nunique()
total_users = traffic['country'].count()
click_events = (traffic['event'] == 'click').sum()
print("Total Sessions:", total_sessions)
print("Total Users:", total_users)
print("Total Click Events:", click_events)
```

```python
traffic['country'].value_counts().head(10).plot(
    kind='bar',
    figsize=(8,5)
)
plt.title('Top 10 Countries')
plt.xlabel('Country')
plt.ylabel('Traffic Count')
plt.show()
```

```python
traffic['city'].value_counts().head(10).plot(
    kind='bar',
    figsize=(8,5)
)
plt.title('Top 10 Cities')
plt.xlabel('City')
plt.ylabel('Traffic Count')
plt.show()
```

```python
traffic['event'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)
plt.title('User Event Distribution')
plt.ylabel('')
plt.show()
```

```python
traffic['track'].value_counts().head(10).plot(
    kind='barh',
    figsize=(8,5)
)
plt.title('Most Visited Pages')
plt.xlabel('Visits')
plt.ylabel('Pages')
plt.show()
```

```python
daily_visits = traffic['date'].value_counts().sort_index()
daily_visits.plot(figsize=(10,5))
plt.title('Daily Website Traffic')
plt.xlabel('Date')
plt.ylabel('Visits')
plt.show()
```

Insights

1. Website traffic was generated from multiple countries and cities.

2. Some pages received significantly higher visits than others.

3. User interactions were mainly driven by click events.

4. Traffic varied across different dates, indicating changing user activity.

5. Data cleaning improved the quality and reliability of the analysis.
