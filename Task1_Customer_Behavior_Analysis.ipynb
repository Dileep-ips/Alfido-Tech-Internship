```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customer_df = pd.read_csv('/content/ecommerce_customer_data_large.csv')

customer_df.head()
```

```python
customer_df.info()
```

```python
customer_df.isnull().sum()
```

```python
customer_df['Returns'] = customer_df['Returns'].fillna(0)
```

```python
customer_df.drop_duplicates(inplace=True)

customer_df.shape
```

```python
customer_df.describe()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x='Gender',
    data=customer_df,
    palette='Set2'
)

plt.title('Gender Distribution of Customers')

plt.show()
```

```python
category_data = customer_df['Product Category'].value_counts()

plt.figure(figsize=(7,7))

category_data.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Customer Product Category Preferences')

plt.ylabel('')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='Payment Method',
    y='Total Purchase Amount',
    data=customer_df
)

plt.title('Customer Spending by Payment Method')

plt.xticks(rotation=15)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    customer_df['Age'],
    bins=15,
    kde=True
)

plt.title('Customer Age Analysis')

plt.xlabel('Age')

plt.ylabel('Frequency')

plt.show()
```

```python
print("Insights")
print("1. Customers preferred specific product categories more frequently.")
print("2. Spending behavior differed across payment methods.")
print("3. Most customers belonged to middle age groups.")
print("4. Data preprocessing improved analysis quality.")
print("5. Visualizations helped identify customer trends.")
```
